```python
data_emp = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}
df_emp = pd.DataFrame(data_emp)
print("\nQ.6a Shape of DataFrame:", df_emp.shape)
print("\nQ.6b Summary:")
df_emp.info()
print("\nQ.6c Descriptive statistics:\n", df_emp.describe())
print("\nQ.6d First 5 rows:\n", df_emp.head())
print("\nQ.6d Last 3 rows:\n", df_emp.tail(3))
print("\nQ.6e Statistics:")
print("i. Average Salary:", df_emp['Salary'].mean())
print("ii. Total Bonus:", df_emp['Bonus'].sum())
print("iii. Youngest Age:", df_emp['Age'].min())
print("iv. Highest Rating:", df_emp['Rating'].max())
df_emp_sorted = df_emp.sort_values(by='Salary', ascending=False)
print("\nQ.6f Sorted by Salary (Descending):\n", df_emp_sorted[['Name', 'Salary']])
def categorize_rating(rating):
    if rating >= 4.5:
        return 'Excellent'
    elif rating >= 4.0:
        return 'Good'
    else:
        return 'Average'
df_emp['Performance_Category'] = df_emp['Rating'].apply(categorize_rating)
print("\nQ.6g Added Performance Category:\n", df_emp[['Name', 'Rating', 'Performance_Category']])
print("\nQ.6h Missing values:\n", df_emp.isnull().sum())
df_emp.rename(columns={'Employee_ID': 'ID'}, inplace=True)
print("\nQ.6i Renamed Columns:\n", df_emp.columns.tolist())
mask = (df_emp['Years_of_Experience'] > 5) & (df_emp['Department'] == 'IT')
print("\nQ.6j IT Employees with >5 years experience:\n", df_emp[mask])
df_emp['Tax'] = df_emp['Salary'] * 0.10
print("\nQ.6k Added Tax column:\n", df_emp[['Name', 'Salary', 'Tax']])
df_emp.to_csv('employees_modified.csv', index=False)
print("\nQ.6l The dataset has been saved successfully to 'employees_modified.csv'.")
```
