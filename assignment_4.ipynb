```python
import pandas as pd

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee", "category": "billing"},
]

custom_entries = [
    {"question": "how do i update my registered mobile number",
     "answer": "Go to Account Settings and update your registered mobile number.",
     "keywords": "mobile number update account", "category": "account"},
    {"question": "how do i change my registered email address",
     "answer": "Open Account Settings, select Email, and enter your new email address.",
     "keywords": "email change update account", "category": "account"},
]

final_entries = fixed_entries + custom_entries

df = pd.DataFrame(final_entries)

print(df)
```

```python
import re
import pandas as pd

def score_query(query, df):
    query_words = set(re.findall(r'\w+', query.lower()))
    results = []

    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(re.findall(r'\w+', text))
        matches = query_words.intersection(entry_words)

        if matches:
            confidence = len(matches) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": round(confidence, 2)
            })
    results.sort(key=lambda x: x["confidence"], reverse=True)

    return pd.DataFrame(results)
query = "how can i update my account mobile number"

results = score_query(query, df)

print(results)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"].tolist()
category_name = "account"

result = same_category(category_name, df)

print(result)
```
