```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
temp_corrected = temperature + 2
print("a. Corrected temperatures (+2 deg C):", temp_corrected)
temp_fahrenheit = (9 / 5) * temp_corrected + 32
print("b. Temperatures in Fahrenheit:", temp_fahrenheit)
above_32 = temp_corrected[temp_corrected > 32]
print("c. Readings greater than 32 deg C:", above_32)
count_above_32 = np.sum(temp_corrected > 32)
print("d. Count of readings exceeding 32 deg C:", count_above_32)

# e. Efficiency explanation
print("Vectorization and Boolean indexing execute contiguous operations directly in C-compiled memory blocks.")
print("This eliminates Python interpreter loop overhead, dynamic type-checking, and memory allocation delay.")
```

**Q2**

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("a. Total steps recorded:", np.sum(steps))
print("b. Mean steps:", np.mean(steps))
print("c. Max steps:", np.max(steps), "| Min steps:", np.min(steps))
print("d. Total steps per day (Day 1, Day 2, Day 3):", np.sum(steps, axis=0))
print("e. Total steps per user (User 1 to 4):", np.sum(steps, axis=1))
flat_max_idx = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_max_idx, steps.shape)
print("f. Argmax flat index:", flat_max_idx)
print("   Max steps position -> User Index:", user_idx, "(User", user_idx + 1, "), Day Index:", day_idx, "(Day", day_idx + 1, ")")
```

**Q4**

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("a. Matrix X - Shape:", X.shape, "| Dimensions:", X.ndim)

X_T = X.T
print("b. X Transposed (X.T):\n", X_T)
print("   Explanation: Transposing swaps rows and columns. X changes from 4 users x 3 features to 3 features x 4 users.")

XTX = X_T @ X
print("\nc. Product X.T @ X:\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nd. Inverse of (X.T @ X):\n", XTX_inv)

beta = XTX_inv @ X_T @ y
print("\ne. Calculated Coefficients (Beta):", beta)

print("\nf. Coefficient Explanation:")
print("   - Beta[0] (Sleep hours):", beta[0], "| Impact of 1 additional hour of sleep on assistance score.")
print("   - Beta[1] (Activity level):", beta[1], "| Impact of 1 unit increase in activity level.")
print("   - Beta[2] (Stress level):", beta[2], "| Impact of 1 unit increase in stress level.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\ng. New User features [5, 40, 7]:")
print("   Predicted Assistance Score:", predicted_score)
```
