```python
roll_number = "1024170277"
L = [int(digit) * 10 for digit in roll_number]
print("i. L:", L)
L.append(99)
L.insert(2, 55)
print("ii. :", L)
L.remove(55)
popped_value = L.pop(2)
print("iii. popped_value: ", L)
L.sort()
print("iv. Sorted ascending:", L)
L.sort(reverse=True)
print("iv. Sorted descending:", L)
print("v. First three elements:", L[:3])
print("v. Last three elements:", L[-3:])
avg_L = sum(L) / len(L)
greater_than_avg = [x for x in L if x > avg_L]
print("vi. Elements greater than average : ", greater_than_avg)
```

**Q3**

```python
import random
roll_number = "1024170277"
random.seed(int(roll_number))
random_numbers = [random.randint(100, 900) for i in range(100)]
odd_numbers = [x for x in random_numbers if x % 2 != 0]
print("\nii. Odd numbers count:", len(odd_numbers), "\n", odd_numbers)
even_numbers = [x for x in random_numbers if x % 2 == 0]
print("iii. Even numbers count:", len(even_numbers), "\n", even_numbers)
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True
prime_numbers = [x for x in random_numbers if is_prime(x)]
print("iv. Prime numbers count:", len(prime_numbers), "\n", prime_numbers)
most_frequent = max(set(random_numbers), key=random_numbers.count)
print("v. Most frequent number:", most_frequent, "(occurs", random_numbers.count(most_frequent), "times)")
```
