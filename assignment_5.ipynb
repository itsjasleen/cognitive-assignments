**Q2**

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("a) Reversed array:", arr[::-1])
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print("b.i) Most frequent in x is", most_freq_x, "at indices", indices_x)
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_freq_y = vals_y[counts_y == max_count_y]
indices_y = np.where(np.isin(y, most_freq_y))[0]
print("b.ii) Most frequent in y are", most_freq_y, "at indices", indices_y)
```

**Q4**

```python
kushagra = np.linspace(10, 100, 25)
print("Dimensions :", kushagra.ndim)
print("Shape:", kushagra.shape)
print("Size :", kushagra.size)
print("Data type :", kushagra.dtype)
print("Total bytes consumed:", kushagra.nbytes)
transpose = kushagra.reshape(25, 1)
print("\nTranspose :\n", transpose)
```

**Q5**

```python
ucs420_kushagra = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("Original 3x4 Array:\n", ucs420_kushagra)

print("\nMean:", np.mean(ucs420_kushagra))
print("Median:", np.median(ucs420_kushagra))
print("Max:", np.max(ucs420_kushagra))
print("Min:", np.min(ucs420_kushagra))
print("Unique elements:", np.unique(ucs420_kushagra))

reshaped_ucs420_kushagra = ucs420_kushagra.reshape(4, 3)
print("\nReshaped to 4x3:\n", reshaped_ucs420_kushagra)

resized_ucs420_kushagra = np.resize(ucs420_kushagra, (2, 3))
print("\nResized to 2x3:\n", resized_ucs420_kushagra)
```
